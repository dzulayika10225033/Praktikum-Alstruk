```python
# Program Penghitung Nilai Akhir Mahasiswa

# 1. Input nama mahasiswa
nama = input("Masukkan nama mahasiswa: ")

# 2. Input nilai tugas, UTS, dan UAS
nilai_tugas = float(input("Masukkan nilai Tugas: "))
nilai_uts = float(input("Masukkan nilai UTS: "))
nilai_uas = float(input("Masukkan nilai UAS: "))

# 3. Hitung nilai akhir dengan bobot Tugas 30%, UTS 30%, UAS 40%
nilai_akhir = (nilai_tugas * 0.30) + (nilai_uts * 0.30) + (nilai_uas * 0.40)

# 4. Tentukan status kelulusan
if nilai_akhir >= 60:
    status = "Lulus"
else:
    status = "Belum Lulus"

# 5. Tampilkan hasil
print("\n===== HASIL PENILAIAN =====")
print("Nama Mahasiswa :", nama)
print("Nilai Akhir    :", round(nilai_akhir, 2))
print("Status         :", status)
```
