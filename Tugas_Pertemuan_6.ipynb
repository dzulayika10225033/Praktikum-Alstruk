```python
#Program Kasir Cafe

print("=== KASIR CAFE ===")
print("1. Nasi Goreng  - Rp12.000")
print("2. Mie Ayam     - Rp10.000")
print("3. Es Teh       - Rp3.000")
print("4. Air Mineral  - Rp2.000")
pilihan = int(input("Pilih menu (1-4): "))
jumlah = int(input("Jumlah pesanan: "))

#menentukan nama dan harga menu
if pilihan == 1:
    menu, harga = "Nasi Goreng", 12000
elif pilihan == 2:
    menu, harga = "Mie Ayam", 10000
elif pilihan == 3:
    menu, harga = "Es Teh", 3000
elif pilihan == 4:
    menu, harga = "Air Mineral", 2000
else:
    menu, harga = None, 0

if menu is None:
    print("Menu tidak tersedia!")
else:
    # NESTED IF : jumlah harus valid sebelum memeriksa pembayaran
    if jumlah > 0:
        total = harga * jumlah
        print(f"\nPesanan : {menu} x {jumlah}")
        print(f"Total   : Rp{total:,}")
        bayar = int(input("Uang bayar (Rp): "))

        #memeriksa uang pembayaran
        if bayar < total:
            kurang = total - bayar
            print(f"Uang kurang Rp{kurang:,}. Pembayaran gagal.")
        elif bayar == total:
            print("Uang pas. Terima kasih!")
        else:
            kembalian = bayar - total
            print(f"Kembalian: Rp{kembalian:,}. Terima kasih!")
    else:
        print("Jumlah pesanan tidak valid!")
```
