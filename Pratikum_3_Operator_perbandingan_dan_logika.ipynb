```python
#operator and
nilai = 82
kehadiran = 90

syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

print("Nilai      :",syarat_nilai)
print("Kehadiran  :",syarat_kehadiran)
print("LULUS      :",syarat_nilai and syarat_kehadiran)
```
