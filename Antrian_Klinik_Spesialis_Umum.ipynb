```python

dokter = ["dr. Ani", "dr. Budi", "dr. Citra"]

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

head = None
nomor = 0

def daftar(nama, keluhan):
    global head, nomor
    nomor += 1
    baru = Node((nomor, nama, keluhan))
    if head is None:
        head = baru
    else:
        cur = head
        while cur.next:
            cur = cur.next
        cur.next = baru
    print(f"{nama} terdaftar dengan nomor antrean {nomor}")

def panggil():
    global head
    if head is None:
        return print("Antrean kosong")
    no, nama, keluhan = head.data
    dr = dokter[(no - 1) % len(dokter)]
    print(f"Antrean {no}: {nama} ({keluhan}) -> {dr}")
    head = head.next

while True:
    print("\n1. Daftar pasien  2. Panggil pasien  0. Keluar")
    pilih = input("Pilih: ")
    if pilih == "1":
        nama = input("Nama    : ")
        keluhan = input("Keluhan: ")
        daftar(nama, keluhan)
    elif pilih == "2":
        panggil()
    elif pilih == "0":
        break
```
