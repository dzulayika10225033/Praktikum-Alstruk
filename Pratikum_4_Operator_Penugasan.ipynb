```python
poin = 100
print("Nilai awal :",poin)

poin += 25
print("Poin += 25 :",poin)
poin -= 10
print("Poin -= 10 :",poin)
poin *= 2
print("Poin *= 2 :",poin)
poin /= 2
print("Poin /= 2 :",poin)
```
